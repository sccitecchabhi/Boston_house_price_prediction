```python
# Checking outliers
df.select_dtypes(include='number').plot(
    kind='box',
    subplots=True,
    layout = (-1,3),
    figsize=(15,8)
)
plt.show()
```

```python
# check actual outliers
def detect_out():
    data = df.select_dtypes(include='number')

    for i in data:
        data=df[i].dropna()
        Q1 = data.quantile(0.25)
        Q3 = data.quantile(0.75)

        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = data[(data < lower) | (data > upper)]

        print("\ncolumn", i)
        print("no_outlier :", len(outliers))
        
        if len(outliers)>0:
            print(outliers)
detect_out()
```
