### Importing the necessary libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor


from sklearn.ensemble import AdaBoostRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.ensemble import AdaBoostRegressor


from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    root_mean_squared_error,
    r2_score
)


import joblib
```

### Load dataset 

```python
df = pd.read_csv('Clean_Data_Boston.csv')
```

```python
df.head()
```

### Splitting the data for input and output 

```python
x = df.drop(columns = 'medv') #Spliting the inputs 
x.head()

y = df['medv'] #Spliting the output 
y.head()
```

```python
x.head()
```

```python
num_col = x.columns
```

### Splitting the data for the training and tsting 

```python
X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size=0.2)
```

```python
X_train.shape,X_test.shape,Y_train.shape,Y_test.shape
```

### Transformation 

```python
num_features = x.select_dtypes(include=["int64", "float64"]).columns
cat_features = x.select_dtypes(include=["object"]).columns

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), num_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), cat_features)
    ]
)
```

### Training Models 

### Linear Regression 

```python
pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", LinearRegression())
])
```

```python
#Training the model 
pipeline.fit(X_train,Y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

```python
metrics = {
    "MAE": mean_absolute_error(Y_test, y_pred),
    "MSE": mean_squared_error(Y_test, y_pred),
    "RMSE": root_mean_squared_error(Y_test, y_pred),
    "R2_Score": r2_score(Y_test, y_pred)
}
```

```python

joblib.dump(metrics, "Linear_Metrics.pkl")
```

```python
print(metrics)
```

```python

mae = mean_absolute_error(Y_test, y_pred)
mse = mean_squared_error(Y_test, y_pred)
rmse = root_mean_squared_error(Y_test, y_pred)
r2 = r2_score(Y_test, y_pred)

print(f"MAE  : {mae:.2f}")
print(f"MSE  : {mse:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R2   : {r2:.4f}")
```
