```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,LabelEncoder,OneHotEncoder
from sklearn.linear_model import LinearRegression 
from sklearn.metrics import accuracy_score,mean_absolute_error,mean_squared_error,r2_score,classification_report
import joblib
```

```python
df=pd.read_csv("boston_data.csv")
df
```

```python
ndf=df.copy()
```

```python
# | **Column**  | **Full Form / Meaning**                 | **Simple Example**                              |
# | ----------- | --------------------------------------- | ----------------------------------------------- |
# | **ID**      | Unique house ID                         | House No. 101                                   |
# | **indus**   | Percentage of industrial land           | 12 = Area has 12% industrial land               |
# | **chas**    | Charles River (0 = No, 1 = Yes)         | 1 = House is near Charles River                 |
# | **nox**     | Nitric Oxide pollution level            | Higher value = More air pollution               |
# | **rm**      | Average number of rooms                 | 6.5 = Average 6.5 rooms per house               |
# | **age**     | % of houses built before 1940           | 80 = 80% houses are old                         |
# | **dis**     | Weighted distance to employment centres | Higher value = House is farther from workplaces |
# | **rad**     | Accessibility to highways               | 1 = Poor access, 24 = Excellent access          |
# | **tax**     | Property tax rate                       | 300 = Property tax index/rate                   |
# | **ptratio** | Pupil-Teacher Ratio                     | 18 = 18 students for every 1 teacher            |
# | **lstat**   | % of lower socioeconomic population     | 10 = 10% lower-income population                |
# | **medv**    | Median house value (Target)             | 25 = $25,000 (values are in $1,000s)            |
```

```python
# Split feture and target
# Features
x = df.drop(columns=["ID","medv"])

# Target
y = df["medv"]
```

```python
# Split training and testing data
X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size=0.2)
```

```python
X_train.shape,X_test.shape,Y_train.shape,Y_test.shape
```

```python
# Separate categorical and numerical columns
cat = []
num=[]

for i in x.columns:
    if df[i].dtypes in ['category','object','str']:
        cat.append(i)
    else:
        num.append(i)
print(cat)
print(num)
```

```python
# Preprocessing
# Labelencoder not work and here One hot Gives sparse matrix and naive want each column mean,variance etc so usesprase_out = False
preprocessor = ColumnTransformer(
      transformers=[
        # we don't use any encode bcz there is already all in numerical and having no categorical columns
          ("num",StandardScaler(),num)
      ])
```

```python
pipeline = Pipeline([
    ("preprocessor",preprocessor),
    ("model",LinearRegression()),
])
```

```python
pipeline.fit(X_train,Y_train)
```

```python
y_pred_train = pipeline.predict(X_train)
```

```python
# Check Training accuracy
print("MAE_train",mean_absolute_error(y_pred_train, Y_train))
print("MSE_train",mean_squared_error(y_pred_train, Y_train))
print("r2_train",r2_score(y_pred_train, Y_train))
```

```python
y_pred_test = pipeline.predict(X_test)
```

```python
# Check Testing accuracy
print("MAE_test",mean_absolute_error(y_pred_test, Y_test))
print("MSE_test",mean_squared_error(y_pred_test, Y_test))
print("r2_test",r2_score(y_pred_test, Y_test))
```

```python
joblib.dump(pipeline,"churn_linearregression.pkl")
```
